# Разведочный анализ данных (EDA)

Основные задачи:

- **Оценить качество данных.** Изучить типы признаков, пропуски, дубликаты и потенциальные выбросы.
- **Исследовать временную и пространственную структуру.** Определить доступные периоды и территории, оценить полноту покрытия и различия между секторами.
- **Проанализировать целевую переменную.** Изучить её распределение, динамику во времени, сезонность и различия между территориями.
- **Исследовать взаимосвязи между признаками.** Оценить зависимости между показателями и определить потенциально информативные группы признаков.
- **Визуализировать основные закономерности.** Подготовить графики, отражающие динамику целевой переменной, пространственные различия и ключевые зависимости в данных.
- **Проверить возможные утечки данных.** Убедиться, что признаки, используемые для прогнозирования, доступны на момент формирования прогноза.
- **Подготовить рабочую версию датасета.** Выполнить базовую очистку и объединение исходных таблиц для дальнейшего моделирования.

## Подготавливаем исходную таблицу с целью

Целевая переменная — `amount_new_house_transactions`, суммарная стоимость зарегистрированных сделок с новостройками в секторе за месяц. В исходной таблице присутствуют только строки с положительным объёмом. Правило интерпретации отсутствующих комбинаций «месяц × сектор» в предоставленных файлах не зафиксировано, поэтому в EDA не будем автоматически считать их нулевыми сделками.

```python
TARGET = "amount_new_house_transactions"

df_target_raw = df_nht.copy()
df_target_raw["month_dt"] = pd.to_datetime(df_target_raw["month"], format="%Y-%b")
```

### Формируем полную панель наблюдений

Создадим сетку из всех исторических месяцев и 96 секторов. После присоединения исходной цели пропуск будет означать, что в предоставленной целевой таблице нет записи по этому сочетанию. Его не следует интерпретировать как нулевой объём без подтверждения из документации источника.

```python
# Полный временной диапазон обучающей выборки
all_months = pd.date_range(
    start=df_target_raw["month_dt"].min(),
    end=df_target_raw["month_dt"].max(),
    freq="MS"
)
all_sectors = [f"sector {number}" for number in range(1, 97)]
panel_keys = ["month_dt", "sector"]

# Полная сетка нужна, чтобы отделить отсутствующую запись от числового нуля.
full_grid = pd.MultiIndex.from_product(
    [all_months, all_sectors],
    names=panel_keys
).to_frame(index=False)

df = full_grid.merge(
    df_target_raw[panel_keys + [TARGET]],
    on=panel_keys,
    how="left",
    validate="one_to_one"
)
df["target_observed"] = df[TARGET].notna()

print("Размер полной сетки:", df.shape)
print("Количество месяцев:", df["month_dt"].nunique())
print("Количество секторов:", df["sector"].nunique())
print("Записанных наблюдений цели:", df["target_observed"].sum())
print("Сочетаний без строки в цели:", (~df["target_observed"]).sum())
```

Проверим покрытие цели и наличие секторов, которые не представлены в исходной таблице.

```python
df['sector'].sort_values().unique()
```

Панель содержит все 96 секторов. При этом покрытие исходной цели неполное. Отсутствующая строка сохранена как `NaN`, поэтому дальнейшие описательные расчёты цели выполняются только для наблюдаемых значений.

Отсутствующие записи цели сохраняем как `NaN`. Это отделяет непокрытые сочетания «месяц × сектор» от наблюдаемых сделок и не приписывает отсутствующей записи экономический смысл нулевого спроса.

```python
# Индикатор позволяет учитывать покрытие отдельно от величины цели.
target_coverage = (
    df.groupby("month_dt")
    .agg(
        observed_rows=("target_observed", "sum"),
        missing_rows=("target_observed", lambda values: (~values).sum()),
        observed_share=("target_observed", "mean")
    )
    .reset_index()
)

print("Размер полной панели:", df.shape)
print("Доля записанных целей:", round(df["target_observed"].mean(), 3))
display(target_coverage.head())
```

### Базовая статистика целевой переменной

Оценим масштаб наблюдаемых значений цели, покрытие панели и динамику зарегистрированного объёма сделок.

```python
observed_target = df.loc[df["target_observed"]].copy()

# Базовая статистика цели
display(observed_target[TARGET].describe())

monthly_target = (
    df.groupby("month_dt")
    .agg(
        total_amount=(TARGET, lambda values: values.sum(min_count=1)),
        mean_amount=(TARGET, "mean"),
        observed_share=("target_observed", "mean")
    )
    .reset_index()
)

print("Записанных наблюдений:", len(observed_target))
print("Непокрытых сочетаний месяц × сектор:", (~df["target_observed"]).sum())
display(monthly_target.head())
```

Результат показывает особенности наблюдаемой части цели:

- Полная панель включает 96 секторов и 67 месяцев, всего 6432 сочетания.
- В исходной таблице записаны 5433 сочетания, или около 84,5% панели. Для оставшихся 999 сочетаний отсутствует строка, и их экономический смысл требует подтверждения в документации источника.
- Среди записанных целей нет нулевых и отрицательных значений. Среднее составляет около 32 588, а медиана — около 16 040 в единицах исходного файла, то есть в десятках тысяч юаней.
- Максимальный записанный объём равен примерно 606 408, или 6,06 млрд юаней. Он существенно превышает типичные значения и создаёт правостороннюю асимметрию.

```python
fig, axes = plt.subplots(2, 1, figsize=(14, 8), sharex=True)

sns.lineplot(data=monthly_target, x="month_dt", y="total_amount", ax=axes[0])
axes[0].set_title("Суммарная стоимость зарегистрированных сделок с новостройками")
axes[0].set_xlabel("")
axes[0].set_ylabel("Сумма, 10 тыс. юаней")

sns.lineplot(data=monthly_target, x="month_dt", y="observed_share", ax=axes[1], color="tomato")
axes[1].set_title("Доля секторов с записанной целевой переменной")
axes[1].set_xlabel("Месяц")
axes[1].set_ylabel("Доля")

plt.tight_layout()
plt.show()
```

Суммарная стоимость зарегистрированных сделок заметно колеблется от месяца к месяцу. Наиболее высокие значения наблюдаются в конце 2020 года и в конце 2021 года, однако устойчивого тренда на рост или снижение за весь период не прослеживается.

Доля строк с записанной целью меняется по месяцам. Без правила источника эти изменения нельзя трактовать как изменения доли секторов без сделок.

```python
target_log = np.log1p(observed_target[TARGET])

plt.figure(figsize=(12, 5))
sns.histplot(target_log, bins=50, kde=True)

plt.title("Распределение логарифма зарегистрированной суммы сделок")
plt.xlabel("log(1 + сумма сделок, 10 тыс. юаней)")
plt.ylabel("Количество наблюдений")
plt.show()
```

Распределение зарегистрированной цели имеет выраженную правостороннюю асимметрию из-за редких крупных сделок. Преобразование `log1p` уменьшает влияние таких значений и делает распределение более компактным. Нулевой столбец не интерпретируется, поскольку нулевые строки в исходной цели отсутствуют, а непокрытые сочетания сохранены как `NaN`.

### Проверка потенциальных выбросов

Оценим верхний хвост распределения через квантили и правило межквартильного размаха. Это позволяет выделить экстремальные наблюдения для проверки, не объявляя их ошибками данных.

```python
target_quantiles = observed_target[TARGET].quantile(
    [0.01, 0.05, 0.25, 0.50, 0.75, 0.95, 0.99]
).to_frame("value")

q1 = target_quantiles.loc[0.25, "value"]
q3 = target_quantiles.loc[0.75, "value"]
upper_iqr_bound = q3 + 1.5 * (q3 - q1)
high_outliers = observed_target[observed_target[TARGET] > upper_iqr_bound]

outlier_summary = pd.DataFrame({
    "value": [upper_iqr_bound, len(high_outliers), len(high_outliers) / len(observed_target)]
}, index=["upper_iqr_bound", "high_outlier_count", "high_outlier_share"])

display(target_quantiles)
display(outlier_summary)
display(high_outliers.nlargest(10, TARGET)[["month_dt", "sector", TARGET]])
```

Наблюдения выше верхней границы IQR образуют верхний хвост распределения. Их не следует автоматически удалять: для рынка недвижимости крупная сделка может быть экономически реальным событием. В модели стоит сравнить подходы без преобразования цели и с `log1p`, а экстремальные строки проверить на корректность исходных значений.

```python
plt.figure(figsize=(12, 2.5))

sns.boxplot(
    x=np.log1p(observed_target[TARGET]),
    color="#8bbadd",
    width=0.5
)

plt.title("Boxplot логарифма зарегистрированной суммы сделок")
plt.xlabel("log(1 + сумма сделок, 10 тыс. юаней)")
plt.show()
```

На логарифмической шкале центральные 50% наблюдений сосредоточены примерно в диапазоне от 8,6 до 10,6, а медиана находится около 9,7. Преобразование сжало исходный правый хвост, поэтому точки за пределами «усов» видны преимущественно слева: это наблюдения с особенно малыми суммами сделок, а не автоматически ошибки данных. График подтверждает, что логарифмирование делает масштаб цели более компактным, но не является основанием удалять наблюдения.

### Проверка согласованности показателей первичного рынка

В таблице `new_house_transactions.csv` несколько признаков описывают одни и те же сделки с разных сторон. Проверим, насколько согласуются между собой количество сделок, площадь, цена и денежный объём.

Такая проверка позволяет выявить потенциально нетипичные наблюдения и понять, какие показатели содержат дублирующую информацию.

```python
df_nht.columns
```

```python
consistency = df_nht.copy()

# Денежный объем через площадь и цену
consistency["amount_calc"] = (
    consistency["area_new_house_transactions"]
    * consistency["price_new_house_transactions"]
    / 10_000
)

# Средняя площадь одной сделки
consistency["area_per_unit_calc"] = (
    consistency["area_new_house_transactions"]
    / consistency["num_new_house_transactions"]
)

# Средняя стоимость одной сделки
consistency["total_price_per_unit_calc"] = (
    consistency["amount_new_house_transactions"]
    / consistency["num_new_house_transactions"]
)

# Средняя стоимость сделки через среднюю площадь и цену за м²
consistency["total_price_per_unit_from_area_price"] = (
    consistency["area_per_unit_new_house_transactions"]
    * consistency["price_new_house_transactions"]
    / 10_000
)


def relative_error(actual, calculated):
    return (
        (calculated - actual).abs()
        / actual.abs().replace(0, np.nan)
    )


consistency["err_amount"] = relative_error(
    consistency["amount_new_house_transactions"],
    consistency["amount_calc"]
)

consistency["err_area_per_unit"] = relative_error(
    consistency["area_per_unit_new_house_transactions"],
    consistency["area_per_unit_calc"]
)

consistency["err_total_price_per_unit"] = relative_error(
    consistency["total_price_per_unit_new_house_transactions"],
    consistency["total_price_per_unit_calc"]
)

consistency["err_unit_price"] = relative_error(
    consistency["total_price_per_unit_new_house_transactions"],
    consistency["total_price_per_unit_from_area_price"]
)
```

```python
error_cols = {
    "Площадь × цена → денежный объем": "err_amount",
    "Площадь / число сделок → средняя площадь": "err_area_per_unit",
    "Денежный объем / число сделок → средняя стоимость": "err_total_price_per_unit",
    "Средняя площадь × цена → средняя стоимость": "err_unit_price",
}

consistency_summary = pd.DataFrame({
    name: {
        "median_error": consistency[col].median(),
        "share_gt_1%": (consistency[col] > 0.01).mean(),
        "share_gt_5%": (consistency[col] > 0.05).mean(),
        "share_gt_10%": (consistency[col] > 0.10).mean(),
        "p99_error": consistency[col].quantile(0.99),
        "max_error": consistency[col].max(),
    }
    for name, col in error_cols.items()
}).T

display(
    consistency_summary.style.format("{:.2%}")
)
```

`median_error` показывает типичное относительное расхождение, а `p99_error` — значение, ниже которого находится 99% наблюдений. Показатели `share_gt_1%`, `share_gt_5%` и `share_gt_10%` отражают долю строк с расхождением выше соответствующего порога.

В большинстве наблюдений показатели хорошо согласуются между собой. Наиболее стабильна связь между средней площадью сделки, ценой квадратного метра и средней стоимостью сделки. Более заметные расхождения встречаются в связях с агрегированными показателями площади, количества и денежного объёма сделок, причём сильные ошибки характерны лишь для небольшой доли наблюдений.

```python
both_anomalies = consistency[
    (consistency["err_amount"] > 0.10)
    & (consistency["err_area_per_unit"] > 0.10)
].copy()

print(
    "Наблюдений с расхождением более 10% сразу в двух проверках:",
    len(both_anomalies)
)

display(
    both_anomalies[
        [
            "month",
            "sector",
            "err_amount",
            "err_area_per_unit",
        ]
    ]
    .sort_values("err_amount", ascending=False)
    .head(5)
)
```

В 36 наблюдениях крупное расхождение возникает одновременно в двух независимых проверках: для денежного объёма сделок и для средней площади сделки. Это делает такие строки более подозрительными с точки зрения качества данных, чем единичное отклонение в одном показателе.

При этом аномалии сосредоточены в отдельных секторах и периодах, поэтому перед моделированием их стоит рассматривать отдельно, а не удалять автоматически.

### Вывод

В большинстве наблюдений показатели первичного рынка хорошо согласуются между собой. Наиболее стабильна связь между средней площадью сделки, ценой квадратного метра и средней стоимостью сделки: существенные расхождения практически отсутствуют.

Более заметные отклонения встречаются в связях между агрегированными показателями количества, площади и денежного объёма сделок. В 36 наблюдениях расхождение превышает 10% одновременно для денежного объёма и средней площади сделки.

Такие строки не удаляются автоматически: отклонения могут быть связаны как с особенностями расчёта исходных показателей, так и с качеством данных. Перед моделированием их необходимо учитывать отдельно.

Для оценки связи POI-признаков с суммарной стоимостью сделок используем корреляцию Спирмена. Корреляция Спирмена рассчитывается по рангам, поэтому менее чувствительна к выбросам и позволяет выявить монотонную связь, не предполагая, что она обязательно линейна.
```python
top_correlations = poi_correlations.head(15).sort_values()

plt.figure(figsize=(10, 7))

sns.barplot(
    x=top_correlations.values,
    y=top_correlations.index,
    color="steelblue"
)

plt.axvline(0, color="black", linewidth=1)

plt.title(
    "15 POI-признаков с наибольшей связью "
    "со стоимостью сделок"
)
plt.xlabel("Корреляция Спирмена")
plt.ylabel("Признак")

plt.show()
```

## Объединённая таблица для анализа связей

Отдельный EDA исходных таблиц завершён. Теперь объединим доступные секторные признаки с полной панелью. Цель в `df_eda` остаётся пропущенной для непокрытых сочетаний. Месячные показатели относятся к тому же месяцу, что и наблюдаемая цель, поэтому таблица предназначена для описательного анализа, а не является готовой матрицей признаков для модели.

```python
df_eda = df.copy()

# Статические характеристики секторов.
df_eda = df_eda.merge(
    df_sPOI,
    on="sector",
    how="left",
    validate="many_to_one"
)

# Месячные таблицы уровня «месяц + сектор»
monthly_tables = [
    df_lt,
    df_ltns,
    df_poht,
    df_pohtns,
    df_nhtns
]

for table in monthly_tables:
    table = table.copy()
    table["month_dt"] = pd.to_datetime(table["month"], format="%Y-%b")
    table = table.drop(columns="month")

    df_eda = df_eda.merge(
        table,
        on=["month_dt", "sector"],
        how="left",
        validate="one_to_one"
    )

print("Размер объединённой таблицы:", df_eda.shape)
```

```python
numeric_columns = df_eda.select_dtypes(include="number").columns

target_correlations = (
    df_eda[numeric_columns]
    .corr(method="spearman")[TARGET]
    .drop(TARGET)
    .dropna()
    .sort_values(key=abs, ascending=False)
)

display(target_correlations.head(20))
```

### Проверка временного сдвига признаков

Для оценки связей используем корреляцию Спирмена. Цель и большинство рыночных показателей имеют сильно скошенные распределения с редкими крупными сделками. Спирмен рассчитывается по рангам, поэтому менее чувствителен к выбросам и позволяет выявить монотонные, в том числе нелинейные, зависимости.

Корреляции для значений того же месяца описывают совместное изменение признаков и цели, но не отражают прогностическую ценность. Поэтому отдельно проверим динамические признаки, сдвинутые на 1–3 месяца внутри каждого сектора. Статические POI в этот анализ не включаются, поскольку дата их измерения не указана.


```python
# Для лагов используем только месячные признаки, меняющиеся во времени.
dynamic_features = [
    feature
    for feature in target_correlations.index
    if feature not in df_sPOI.columns
]

lag_panel = df_eda.sort_values(["sector", "month_dt"]).reset_index(drop=True)
lag_results = {}
for lag in [0, 1, 2, 3]:
    shifted_features = lag_panel.groupby("sector", sort=False)[dynamic_features].shift(lag)
    lag_results[lag] = shifted_features.corrwith(
        lag_panel[TARGET],
        method="spearman"
    )

lag_corr_df = pd.DataFrame(lag_results).rename(
    columns=lambda lag: f"lag_{lag}"
)
lag_corr_df["abs_lag_1"] = lag_corr_df["lag_1"].abs()
lag_corr_df = lag_corr_df.sort_values("abs_lag_1", ascending=False)

display(lag_corr_df.head(20).round(3))
```

```python
top_lag_features = lag_corr_df.head(15).index.tolist()

plot_df = (
    lag_corr_df
    .loc[top_lag_features, ["lag_0", "lag_1"]]
    .reset_index()
    .rename(columns={"index": "feature"})
    .melt(id_vars="feature", var_name="тип", value_name="corr")
)

plot_df["feature_short"] = plot_df["feature"].str.replace(
    "_pre_owned_house_transactions", "_pre_owned", regex=False
).str.replace(
    "_new_house_transactions_nearby_sectors", "_new_house_nearby", regex=False
).str.replace(
    "_new_house_transactions", "_new_house", regex=False
)

plt.figure(figsize=(11, 8))
sns.barplot(
    data=plot_df,
    y="feature_short",
    x="corr",
    hue="тип",
    order=plot_df.loc[plot_df["тип"] == "lag_1"]
        .sort_values("corr", key=abs, ascending=False)["feature_short"],
)
plt.axvline(0, color="black", linewidth=1)
plt.title("Одновременная и лаговая (1 месяц) корреляция Спирмена с целью")
plt.xlabel("Корреляция Спирмена")
plt.ylabel("Признак")
plt.legend(title="")
plt.tight_layout()
plt.show()
```

У показателей вторичного рынка и новостроек в соседних секторах после сдвига на один месяц связь с целью ослабевает, но сохраняется. Это делает лаговые версии этих признаков кандидатами для модели.

POI не участвовали в лаговом анализе, поскольку дата их измерения неизвестна. Их можно включать в историческую валидацию только после подтверждения, что они были доступны на соответствующую дату.

```python
# топ-15 признаков по корреляции с целью
top_features = target_correlations.head(15).index.tolist()

# Добавим важные признаки, исключив дубликаты
extra_features = ['area_pre_owned_house_transactions', 'num_pre_owned_house_transactions']
top_features = list(dict.fromkeys(top_features + extra_features))

plt.figure(figsize=(12, 10)) # Немного уменьшим размер, т.к. дублей не будет
correlation_matrix = df_eda[top_features + [TARGET]].corr(method='spearman')
sns.heatmap(correlation_matrix, annot=False, cmap='coolwarm', linewidths=.5)
plt.title('Тепловая карта корреляций топ-признаков')
plt.show()
```

### Интерпретация тепловой карты

Три показателя вторичного рынка - площадь, сумма и число сделок - тесно связаны между собой. Они описывают одну рыночную активность в разных единицах, поэтому их одновременное использование в линейной модели может привести к мультиколлинеарности.

Показатели коммерческой инфраструктуры также образуют тесно связанную группу. Количество магазинов, торговых центров, гипермаркетов и их плотности частично дублируют информацию о развитости сектора.

Связи с целевой переменной слабее, чем связи внутри этих групп. Наиболее заметные связи остаются умеренными, поэтому отдельный признак не объясняет объём сделок самостоятельно. Отрицательные ячейки показывают обратное ранговое соизменение и не доказывают причинную связь.


### Пропуски в объединённой таблице

Внутри тематических групп признаки могут быть тесно связаны между собой. Поэтому использовать все числовые признаки одновременно без отбора нецелесообразно. Для линейных моделей особенно важны регуляризация и контроль мультиколлинеарности.

### Взаимосвязь цели с показателями вторичного рынка

Визуализируем три наиболее связанных показателя. Значения одного и того же месяца используются только для описательного EDA. В модели допустимы лишь значения, доступные на дату прогнозирования для соответствующего горизонта.

```python
top_transaction_features = [
    "area_pre_owned_house_transactions",
    "num_pre_owned_house_transactions",
    "amount_pre_owned_house_transactions"
]

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for axis, feature in zip(axes, top_transaction_features):
    plot_data = df_eda[[TARGET, feature]].dropna()

    sns.scatterplot(
        data=plot_data,
        x=np.log1p(plot_data[feature]),
        y=np.log1p(plot_data[TARGET]),
        alpha=0.35,
        ax=axis
    )

    axis.set_title(feature)
    axis.set_xlabel(f"log(1 + {feature})")
    axis.set_ylabel(f"log(1 + {TARGET})")

plt.tight_layout()
plt.show()
```

Точечные диаграммы подтверждают умеренную, но не сильную связь между рынком вторичного жилья и зарегистрированной стоимостью сделок с новостройками. При увеличении площади, количества и суммы сделок на вторичном рынке цель в среднем также растёт, однако разброс значений очень большой.

Точки вблизи нуля по оси X соответствуют месяцам без сделок на вторичном рынке. На графиках не проведены специальные вертикальные или горизонтальные линии. Поскольку непокрытые сочетания цели сохранены как пропуски, нулевые значения по оси Y здесь не интерпретируются.

### Сезонность целевой переменной

Проверим, различается ли активность рынка по календарным месяцам.

```python
seasonality = (
    df_eda.assign(month_number=df_eda["month_dt"].dt.month)
    .groupby("month_number")
    .agg(mean_amount=(TARGET, "mean"), median_amount=(TARGET, "median"), observed_share=("target_observed", "mean"))
    .reset_index()
)

fig, axes = plt.subplots(2, 1, figsize=(12, 8), sharex=True)
sns.barplot(data=seasonality, x="month_number", y="median_amount", ax=axes[0], color="steelblue")
axes[0].set_title("Медианная стоимость зарегистрированных сделок по месяцам года")
axes[0].set_xlabel("")
axes[0].set_ylabel("Медианная сумма, 10 тыс. юаней")
sns.barplot(data=seasonality, x="month_number", y="observed_share", ax=axes[1], color="tomato")
axes[1].set_title("Доля секторов с записанной целевой переменной")
axes[1].set_xlabel("Номер месяца")
axes[1].set_ylabel("Доля")
plt.tight_layout()
plt.show()
```

График указывает на возможную сезонность среди зарегистрированных сделок. Самая высокая медианная стоимость наблюдается в декабре — около 22,6 тыс. единиц исходного файла, или 226,3 млн юаней. Самая низкая медиана наблюдается в феврале — около 9,4 тыс., или 94,3 млн юаней.

Нижний график отражает покрытие исходной цели, а не долю секторов без сделок. Календарный месяц стоит проверить как признак модели, но устойчивость эффекта необходимо подтвердить временной валидацией.

### Ограничения и следующий этап

Корреляции в EDA описывают совместное изменение признаков и цели, но не доказывают причинность. Повторные наблюдения одного сектора зависимы во времени, последний год неполный, а покрытие источников неодинаково. Поэтому качество признаков необходимо проверять временной валидацией: обучаться на прошлом и оцениваться на более поздних месяцах.
